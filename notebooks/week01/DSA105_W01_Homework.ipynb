## Problem

A bookstore sold:

- 14 textbooks at 240 TL each;
- 9 workbooks at 160 TL each.

Follow:

**Problem → IPO → Algorithm → Prediction → Python → Verification → Transfer**


## 1. IPO

Write the input, process, and output in your own words.
1. INPUT
* Number of textbooks sold :14
* Textbook unit price :240 TL
* Number of workbooks sold:9
* Workbook unit price :160 TL

2.PROCESS
* Calculate the total revenue from textbooks (14 * 240)
* Calculate the total revenue from workbooks (9 * 160)
* Calculate the overall total revenue (Textbooks revenue + Workbooks revenue)

3.OUTPUT
*  The total expected revenue



















## 2. Algorithm

Write 5–7 precise steps.

* textbook_unit_price =240
* textbook_quantity = 14
* workbook_unit_price = 160
* workbook_quantity = 9
* textbook_revenue = textbook_unit_price * textbook_quantity
* workbook_revenue = workbook_unit_price * workbook_quantity
* total_revenue = textbook_revenue + workbook_revenue











## 3. Prediction

Calculate the expected revenues and total by hand before coding.
4800 TL

```python
textbook_unit_price = 240
textbook_quantity = 14
workbook_unit_price = 160
workbook_quantity = 9

textbook_revenue = textbook_unit_price * textbook_quantity
workbook_revenue = workbook_unit_price * workbook_quantity
total_revenue = textbook_revenue + workbook_revenue

print(f"Textbook Revenue : {textbook_revenue} TL")
print(f"workbook Revenue : {workbook_revenue} TL")
print(f"Total Revenue : {total_revenue} TL")
# Complete the program using meaningful names.
```
